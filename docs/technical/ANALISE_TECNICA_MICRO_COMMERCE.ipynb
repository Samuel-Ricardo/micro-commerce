### 1.1 Global Exception Handling ✅
**Objetivo**: Centralizar tratamento de erros
**Impacto**: Erros consistentes, melhor observabilidade
**Esforço**: 1 dia

```java
// ✅ api-gateway/src/main/java/com/js/gateway/infra/GlobalExceptionHandler.java
@RestControllerAdvice
@Slf4j
public class GlobalExceptionHandler {
    
    @ExceptionHandler(InsufficientStockException.class)
    public ResponseEntity<ErrorResponse> handleInsufficientStock(
        InsufficientStockException ex) {
        return ResponseEntity
            .status(BAD_REQUEST)
            .body(new ErrorResponse(
                "INSUFFICIENT_STOCK",
                "Product is out of stock",
                ex.getMessage()
            ));
    }
    
    @ExceptionHandler(ServiceUnavailableException.class)
    public ResponseEntity<ErrorResponse> handleServiceUnavailable(
        ServiceUnavailableException ex) {
        return ResponseEntity
            .status(SERVICE_UNAVAILABLE)
            .body(new ErrorResponse(
                "SERVICE_UNAVAILABLE",
                "Service is temporarily unavailable",
                ex.getMessage()
            ));
    }
    
    @ExceptionHandler(Exception.class)
    public ResponseEntity<ErrorResponse> handleGenericException(Exception ex) {
        log.error("Unhandled exception", ex);
        return ResponseEntity
            .status(INTERNAL_SERVER_ERROR)
            .body(new ErrorResponse(
                "INTERNAL_ERROR",
                "An unexpected error occurred",
                null
            ));
    }
}

// Model
@Data @Builder
public class ErrorResponse {
    private String errorCode;
    private String message;
    private String details;
    private LocalDateTime timestamp = LocalDateTime.now();
}
```

**Validation**: 
- [ ] Todas exceções retornam ErrorResponse consistent
- [ ] Logs estruturados com errorCode
- [ ] Testes para cada handler

### 1.2 Circuit Breaker Enhancement ✅
**Objetivo**: Melhorar resiliência com fallback
**Impacto**: Sistema mais tolerante a falhas
**Esforço**: 2 dias

```java
// order-service/src/main/java/com/js/microservices/order_service/infra/ResilienceConfig.java
@Configuration
public class ResilienceConfig {
    
    @Bean
    public Bulkhead inventoryBulkhead() {
        BulkheadConfig config = BulkheadConfig.custom()
            .maxConcurrentCalls(10)
            .maxWaitDuration(Duration.ofSeconds(2))
            .build();
        return Bulkhead.of("inventory", config);
    }
    
    @Bean
    public RateLimiter inventoryRateLimiter() {
        RateLimiterConfig config = RateLimiterConfig.custom()
            .limitRefreshPeriod(Duration.ofMinutes(1))
            .limitForPeriod(100)
            .timeoutDuration(Duration.ofSeconds(5))
            .build();
        return RateLimiter.of("inventory", config);
    }
}

// Gateway fallback melhorado
@Configuration
public class Router {
    
    @Bean
    public RouterFunction<ServerResponse> orderServiceWithResilience() {
        return GatewayRouterFunctions.route("order_service")
                .route(RequestPredicates.path("/api/v1/orders"), http(orderUrl))
                .filter(CircuitBreakerFilterFunctions.circuitBreaker(
                    "orderServiceCB", 
                    URI.create("forward:/fallback/degraded")))
                .filter(RateLimiterFilterFunctions.rateLimiter(
                    "orderServiceRL",
                    URI.create("forward:/fallback/rate-limit")))
                .build();
    }
    
    @Bean
    public RouterFunction<ServerResponse> fallbackRoutes() {
        return GatewayRouterFunctions.route("fallbacks")
                .GET("/fallback/degraded", 
                    request -> ServerResponse
                        .status(SERVICE_UNAVAILABLE)
                        .body(new FallbackResponse(
                            "ORDER_SERVICE_UNAVAILABLE",
                            "Order service is temporarily unavailable. Try again later.")))
                .GET("/fallback/rate-limit",
                    request -> ServerResponse
                        .status(TOO_MANY_REQUESTS)
                        .body(new FallbackResponse(
                            "RATE_LIMIT_EXCEEDED",
                            "Too many requests. Please try again later.")))
                .build();
    }
}
```

**Validation**:
- [ ] Circuit breaker abre com 5+ falhas consecutivas
- [ ] Fallback retorna resposta degradada consistente
- [ ] Rate limiter rejeita após 100 req/min
- [ ] Métricas de CB expostas via Prometheus

### 2.1 Structured Logging com Correlation IDs ✅
**Objetivo**: Rastrear requisições através de serviços
**Impacto**: Debugging facilitado de bugs distribuídos
**Esforço**: 2 dias

```java
// ✅ common/MDCFilter.java
@Component
@Slf4j
public class MDCLoggingFilter implements Filter {
    
    private static final String CORRELATION_ID = "X-Correlation-ID";
    private static final String REQUEST_ID = "X-Request-ID";
    
    @Override
    public void doFilter(ServletRequest request, ServletResponse response,
                        FilterChain chain) throws IOException, ServletException {
        HttpServletRequest httpRequest = (HttpServletRequest) request;
        
        String correlationId = httpRequest.getHeader(CORRELATION_ID);
        if (correlationId == null) {
            correlationId = UUID.randomUUID().toString();
        }
        
        String requestId = UUID.randomUUID().toString();
        
        try {
            MDC.put(CORRELATION_ID, correlationId);
            MDC.put(REQUEST_ID, requestId);
            
            chain.doFilter(request, response);
        } finally {
            MDC.clear();
        }
    }
}

// Uso em logs
@Service @Slf4j
public class OrderServiceImpl {
    
    public Order placeOrder(PlaceOrderDTO dto) {
        log.info("Placing order for customer: {}", dto.getEmail());
        // Correlation ID é incluído automaticamente
        // Log: 2026-07-23 10:30:45 [X-Correlation-ID=abc-123] Placing order
        
        return order;
    }
}
```

**Validation**:
- [ ] Todos logs incluem `[X-Correlation-ID]` e `[X-Request-ID]`
- [ ] IDs propagados entre serviços via headers HTTP
- [ ] IDs propagados em mensagens Kafka
- [ ] Dashboard Grafana mostra traces por correlation ID

```java
// ✅ product-service/src/main/java/CacheConfig.java
@Configuration
@EnableCaching
public class CacheConfig {
    
    @Bean
    public RedisCacheManager cacheManager(RedisConnectionFactory factory) {
        RedisCacheConfiguration config = RedisCacheConfiguration.defaultCacheConfig()
            .entryTtl(Duration.ofHours(1))
            .serializeValuesWith(RedisSerializationContext.SerializationPair
                .fromSerializer(new GenericJackson2JsonRedisSerializer()));
        
        return RedisCacheManager.builder(factory)
            .cacheDefaults(config)
            .build();
    }
}

// ✅ Uso
@Service @AllArgsConstructor
public class ProductServiceImpl {
    
    @Cacheable(value = "products", key = "#id")
    public Product getProductById(String id) {
        return repository.findById(id)
            .orElseThrow(ProductNotFoundException::new);
    }
    
    @Cacheable(value = "all_products")
    public List<Product> getAllProducts() {
        return repository.findAll();
    }
    
    @CacheEvict(value = "all_products", allEntries = true)
    public Product createProduct(CreateProductDTO dto) {
        return repository.save(newProductModel(dto));
    }
}
```

## Phase 4: Reliability (Sprint 7-8) - 2 Semanas

### 4.1 Transactional Outbox Pattern ✅
**Objetivo**: Garantir entrega de eventos Kafka
**Impacto**: Zero perda de eventos
**Esforço**: 3 dias

```java
// ✅ order-service/OutboxEvent.java
@Entity @Table(name = "outbox_events")
@Data @NoArgsConstructor @AllArgsConstructor
public class OutboxEvent {
    
    @Id
    @GeneratedValue(strategy = GenerationType.UUID)
    private UUID id;
    
    @Column(nullable = false)
    private String aggregateId;
    
    @Column(nullable = false)
    private String eventType;
    
    @Lob
    @Column(nullable = false)
    private String payload;
    
    @Column(nullable = false)
    private boolean processed = false;
    
    @CreatedDate
    private LocalDateTime createdAt;
    
    @LastModifiedDate
    private LocalDateTime updatedAt;
}

// ✅ Service com Outbox
@Service @Transactional
public class OrderServiceImpl {
    
    @Autowired
    private OrderRepository orderRepository;
    
    @Autowired
    private OutboxEventRepository outboxRepository;
    
    public Order placeOrder(PlaceOrderDTO dto) {
        // 1. Criar order
        Order order = createOrder(dto);
        Order saved = orderRepository.save(newOrderModel(order));
        
        // 2. Persistir evento no Outbox (mesma transação)
        OutboxEvent event = new OutboxEvent();
        event.setAggregateId(saved.getId());
        event.setEventType("OrderPlaced");
        event.setPayload(mapToJson(saved));
        outboxRepository.save(event);
        
        return saved;
        // Commit automático aqui
    }
}

// ✅ Polling para publicar eventos
@Component @Scheduled(fixedDelay = 1000)
public class OutboxPoller {
    
    @Autowired
    private OutboxEventRepository outboxRepository;
    
    @Autowired
    private KafkaTemplate<String, String> kafkaTemplate;
    
    public void pollAndPublish() {
        List<OutboxEvent> events = outboxRepository.findByProcessedFalse();
        
        events.forEach(event -> {
            try {
                kafkaTemplate.send(event.getEventType(), event.getPayload())
                    .get();  // Aguardar confirmação
                
                event.setProcessed(true);
                outboxRepository.save(event);
                
            } catch (Exception ex) {
                log.error("Failed to publish event: {}", event.getId(), ex);
                // Não atualizar processed, tentar novamente depois
            }
        });
    }
}
```

**Validation**:
- [ ] Outbox table criada e migrada
- [ ] Events persistidos transacionalmente
- [ ] Poller publica corretamente
- [ ] Dead events após 24h são alertados

```java
// ✅ RECOMENDADO
@Component
public class JwtValidator {
    
    private final JwtDecoder jwtDecoder;
    
    public boolean isTokenValid(String token) {
        try {
            Jwt jwt = jwtDecoder.decode(token);
            
            // Validar custom claims
            String scope = jwt.getClaimAsString("scope");
            if (!scope.contains("microservices")) {
                return false;
            }
            
            // Validar expiração
            Instant expiresAt = jwt.getExpiresAt();
            return expiresAt.isAfter(Instant.now());
            
        } catch (JwtException e) {
            log.error("JWT validation failed: {}", e.getMessage());
            return false;
        }
    }
}
```
